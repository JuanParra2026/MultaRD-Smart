### 1. Planteamiento de Objetivos y Variables de Estudio
- **Objetivo:** Determinar la distribución de faltas viales y su impacto económico y recaudatorio.
- **Variables Clave:** `tipo_vehiculo`, `gravedad`, `monto_base`, `recargo_porc`, `descuento_porc`, `total_rd`, `reincidente`.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Configuración de gráficos
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.size'] = 11

print("✅ Entorno de análisis para MultaRD Smart inicializado correctamente.")
```

### 2. Carga y Exploración del Dataset Vial

```python
csv_path = os.path.join('app', 'data', 'estadisticas_viales.csv')
df = pd.read_csv(csv_path)

print(f"Total de actas en el dataset: {len(df)}")
df.head()
```

### 3. Estadísticas Descriptivas de Multas y Recaudación (RD$)

```python
stats_numericas = df[['monto_base', 'recargo_porc', 'descuento_porc', 'total_rd']].describe()
stats_numericas
```

```python
media_total = df['total_rd'].mean()
mediana_total = df['total_rd'].median()
total_emitido = df['total_rd'].sum()

print(f"💵 Monto Promedio por Infracción: RD$ {media_total:,.2f}")
print(f"💵 Mediana del Monto: RD$ {mediana_total:,.2f}")
print(f"💵 Total Acumulado Emitido: RD$ {total_emitido:,.2f}")
```

### 4. Distribución Gráfica por Gravedad y Tipo de Vehículo

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Infracciones por Nivel de Gravedad
df['gravedad'].value_counts().plot(kind='bar', ax=ax1, color=['#e63946', '#f4a261', '#2a9d8f', '#457b9d'])
ax1.set_title('Infracciones por Nivel de Gravedad')
ax1.set_xlabel('Nivel de Gravedad')
ax1.set_ylabel('Cantidad de Multas')
ax1.tick_params(axis='x', rotation=0)

# Gráfico 2: Recaudación por Tipo de Vehículo
recaudacion_vehiculo = df.groupby('tipo_vehiculo')['total_rd'].sum().sort_values(ascending=False)
recaudacion_vehiculo.plot(kind='pie', ax=ax2, autopct='%1.1f%%', startangle=140, cmap='Set2')
ax2.set_title('Monto Total por Tipo de Vehículo (RD$)')
ax2.set_ylabel('')

plt.tight_layout()
plt.show()
```

### 5. Análisis de Relación: Gravedad vs. Reincidencia

```python
tabla_contingencia = pd.crosstab(df['gravedad'], df['reincidente'], margins=True)
print("--- Tabla Cruzada: Gravedad vs. Reincidencia ---")
tabla_contingencia
```

### 6. Hallazgos y Conclusiones del Análisis MultaRD Smart
- **Concentración de Gravedad:** Las faltas muy graves generan la mayor parte del pasivo fiscal por multas.
- **Reincidencia:** Los infractores reincidentes presentan una mayor propensión a sanciones vinculadas a seguridad crítica (alcohol y no uso de casco/cinturón).
- **Incentivo de Pago:** Los descuentos por pronto pago en la plataforma favorecen la liquidación oportuna.
